```python
def check_table_validity(ddl_columns, table_name):
    # Parse DDL into a StructType
    expected_schema = _parse_datatype_string(f"struct<{ddl_columns}>")
    actual_schema = spark.table(table_name).schema

    print(f"Table Name: {table_name}")
    print(f"DDL columns: {len(expected_schema.fields)}")
    print(f"Table columns: {len(actual_schema.fields)}")
    for i, (exp, act) in enumerate(zip(expected_schema.fields, actual_schema.fields)):
        if exp.name != act.name or exp.dataType != act.dataType:
            print(f"Mismatch at position {i}: DDL has ({exp.name}, {exp.dataType}) vs table has ({act.name}, {act.dataType})")
```

```python
for table_name, ddl_columns in public_name_ddl_dict.items():
    check_table_validity(ddl_columns, table_name)
```
