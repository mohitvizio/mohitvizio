```python
%sql
SELECT vc.fk_tvid, vc.session_start, vc.session_end, vc.fk_commercial_id, vc.external_id
, cief.brand_name, cief.title, cief.external_id AS cief_ext_id, vc.tvog_num
FROM (
SELECT fk_tvid, session_start, session_end
, COUNT(DISTINCT NVL(fk_commercial_id,0)) AS comm_id_count
, COUNT(DISTINCT NVL(external_id,'')) AS external_id_count
FROM detection.viewing_commercials_firehose_dedup_cfe_merge
WHERE session_start >= CURRENT_DATE - 4
  AND session_start < CURRENT_DATE - 3
  AND MOD(fk_tvid, 10) = 1
  AND fk_commercial_source_id = 1
GROUP BY 1, 2, 3
) AS x
JOIN detection.viewing_commercials_firehose_dedup_cfe_merge vc
  ON vc.fk_tvid = x.fk_tvid
 AND vc.session_start = x.session_start
 AND vc.session_end = x.session_end
JOIN detection.commercial_id_external_firehose cief
  ON cief.fk_commercial_id = vc.fk_commercial_id
JOIN detection.clients cl
  ON cl.client_id = cief.fk_client_id
 AND LOWER(cl.client_name) IN ('kinetiq', 'springserve-prod')
WHERE x.comm_id_count > 1
  AND x.external_id_count > 1
  AND vc.session_start >= CURRENT_DATE - 4
  AND vc.session_start < CURRENT_DATE - 3
  AND MOD(vc.fk_tvid, 10) = 1
  AND vc.fk_commercial_source_id = 1
ORDER BY vc.fk_tvid, vc.session_start
LIMIT 100
```
