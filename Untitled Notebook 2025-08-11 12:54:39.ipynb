```python
%sql
SELECT vod_stations.vendor_station_id, COUNT(*)
FROM prod.detection.viewing_content_firehose vc
join prod.detection.station_distribution_obfuscation_overwrite vod_stations
  on (vc.fk_station_id = vod_stations.vendor_station_id
 and vod_stations.vendor_name = 'TIVO')
 OR (vc.fk_station_id != vod_stations.vendor_station_id
 AND vc.tms_station_id = vod_stations.vendor_station_id
 and vod_stations.vendor_name = 'TMS')
where 1=1
  and session_start >= '2025-08-10 00:00:00' :: timestamp
  and session_end < '2025-08-11 00:00:00' :: timestamp
GROUP BY 1
```

```python
%sql
SELECT MIN(start_ts), MAX(start_ts) FROM prod.customer_reports.wfd_watermark as wfd_check
WHERE wfd_check.report_type = 'commercial'
  AND wfd_check.frequency = 'hourly'
    -- # AND mt.session_start_hour = date_format(wfd_check.start_ts, 'yyyy-MM-dd HH:mm:ss')
```

```python
%sql
SELECT DATE(mt.session_start), COUNT(*)
FROM prod.detection.viewing_commercials_golden AS mt
LEFT JOIN prod.customer_reports.wfd_watermark as wfd_check
    ON wfd_check.report_type = 'commercial'
    AND wfd_check.frequency = 'hourly'
    AND mt.session_start_hour = wfd_check.start_ts
WHERE MOD(mt.fk_tvid, 100) = 1
AND (
  (mt.session_start between (current_date -  INTERVAL '1' MONTH) AND current_date)
  or wfd_check.report_name is not null
  )
GROUP BY 1
```
