```python
%sql
CREATE TABLE dev.mohit_gangwani.dp4_dp5_individual_examples_thousanth_of_pop AS
(
  SELECT tvid
  , ts_start
  , ts_end
  , ts_duration
  , cid
  , chan_callsign
  , epid
  , 'DP4' AS pipeline
  FROM (
    SELECT *
    , LEAD(ts_start) OVER (PARTITION BY tvid ORDER BY ts_start, ts_end) AS next_start
    , LAG(ts_end) OVER (PARTITION BY tvid ORDER BY ts_start, ts_end) AS prev_end
    , LEAD(ts_duration) OVER (PARTITION BY tvid ORDER BY ts_start, ts_end) AS next_duration
    , LAG(ts_duration) OVER (PARTITION BY tvid ORDER BY ts_start, ts_end) AS prev_duration
    , CASE WHEN next_start < ts_end AND DATE_TRUNC('HOUR', next_start) = DATE_TRUNC('HOUR', ts_end) THEN 1
           WHEN prev_end > ts_start AND DATE_TRUNC('HOUR', prev_end) = DATE_TRUNC('HOUR', ts_start) THEN 1
      END AS overlapping_session
    , CASE WHEN overlapping_session IS NULL THEN 1
           ELSE CASE WHEN next_start IS NOT NULL AND next_start < ts_end AND next_duration < ts_duration AND DATE_TRUNC('HOUR', next_start) = DATE_TRUNC('HOUR', ts_start) THEN 1
                     WHEN prev_end IS NOT NULL AND prev_end > ts_start AND prev_duration < ts_duration AND DATE_TRUNC('HOUR', prev_end) = DATE_TRUNC('HOUR', ts_end) THEN 1
        END END AS keep
    FROM (
      SELECT tvid
      , ts_start
      , ts_end
      , TIMESTAMPDIFF(SECOND, ts_start, ts_end) AS ts_duration
      , cid
      , chan_callsign
      , epid
      , ROW_NUMBER() OVER (PARTITION BY tvid, ts_start ORDER BY ts_start, ts_end DESC, epid) AS rn
      FROM prod.staging.vizio_content_firehose
      WHERE ts_start>=CURRENT_DATE - INTERVAL 8 DAYS
        AND ts_start < CURRENT_TIMESTAMP - INTERVAL 8 HOURs
        AND MOD(tvid, 1000) = 1
      
    )
    WHERE rn = 1
  ) vc
  WHERE keep = 1
  GROUP BY ALL
)
UNION (
  SELECT tvid
  , ts_start
  , ts_end
  , ts_duration
  , cid
  , chan_callsign
  , epid
  , 'DP5' AS pipeline
  FROM (
    SELECT *
    , LEAD(ts_start) OVER (PARTITION BY tvid ORDER BY ts_start, ts_end) AS next_start
    , LAG(ts_end) OVER (PARTITION BY tvid ORDER BY ts_start, ts_end) AS prev_end
    , LEAD(ts_duration) OVER (PARTITION BY tvid ORDER BY ts_start, ts_end) AS next_duration
    , LAG(ts_duration) OVER (PARTITION BY tvid ORDER BY ts_start, ts_end) AS prev_duration
    , CASE WHEN next_start < ts_end AND DATE_TRUNC('HOUR', next_start) = DATE_TRUNC('HOUR', ts_end) THEN 1
           WHEN prev_end > ts_start AND DATE_TRUNC('HOUR', prev_end) = DATE_TRUNC('HOUR', ts_start) THEN 1
      END AS overlapping_session
    , CASE WHEN overlapping_session IS NULL THEN 1
           ELSE CASE WHEN next_start IS NOT NULL AND next_start < ts_end AND next_duration < ts_duration AND DATE_TRUNC('HOUR', next_start) = DATE_TRUNC('HOUR', ts_start) THEN 1
                     WHEN prev_end IS NOT NULL AND prev_end > ts_start AND prev_duration < ts_duration AND DATE_TRUNC('HOUR', prev_end) = DATE_TRUNC('HOUR', ts_end) THEN 1
        END END AS keep
    FROM (
      SELECT tvid
      , ts_start
      , ts_end
      , TIMESTAMPDIFF(SECOND, ts_start, ts_end) AS ts_duration
      , cid
      , chan_callsign
      , epid
      , ROW_NUMBER() OVER (PARTITION BY tvid, ts_start ORDER BY ts_start, ts_end DESC, epid) AS rn
      FROM prod.cooker.vizio_content_firehose
      WHERE ts_start>=CURRENT_DATE - INTERVAL 8 DAYS
        AND ts_start < CURRENT_TIMESTAMP - INTERVAL 8 HOURs
        AND MOD(tvid, 1000) = 1
    )
    WHERE rn = 1
  ) vc
  WHERE keep = 1
  GROUP BY ALL
)
```

```python
%sql
WITH all_metrics AS (
  SELECT pipeline
  , tvid
  , CASE WHEN ts_start >= '2025-05-11 11:00:00' THEN 'Since Fix' ELSE 'Pre-Fix' END AS time_period
  -- , DATE_TRUNC('HOUR', ts_start) AS session_hour
  , CASE WHEN LOWER(cid) = 'unknown' THEN 'Null Session' ELSE 'Detected Session' END AS session_type
  , SUM(ts_duration)/60.0 AS duration_min
  , COUNT(*) AS session_count
  , SUM(ts_duration)*1.0/COUNT(*) AS avg_session_duration
  FROM dev.mohit_gangwani.dp4_dp5_individual_examples_thousanth_of_pop
  WHERE ts_start >= '2025-05-06 13:00:00'
  GROUP BY 1, 2, 3, 4
)
, detection_rate AS (
  WITH detected_sessions AS (
    SELECT tvid, pipeline, time_period, duration_min
    FROM all_metrics
    WHERE session_type = 'Detected Session'
  )
  , all_stats AS (
    SELECT tvid, pipeline, time_period, SUM(duration_min) AS duration_min
    FROM all_metrics
    GROUP BY 1, 2, 3
  )
  SELECT a.tvid AS tvid
  , a.time_period AS time_period
  , a.pipeline
  , GREATEST(d.duration_min, 0.0)/a.duration_min AS detection_rate
  FROM all_stats a
  LEFT JOIN detected_sessions d
    ON a.tvid = d.tvid
    AND a.time_period = d.time_period
    AND a.pipeline = d.pipeline
  WHERE GREATEST(d.duration_min, 0.0)/a.duration_min > 0
)
, both_tp_filter AS (
  SELECT tvid, pipeline
  FROM detection_rate
  GROUP BY 1, 2
  HAVING COUNT(DISTINCT time_period) = 2
)
SELECT dp4.tvid
, dp4.time_period
, dp4.session_type
, dp4.duration_min AS dp4_duration
, dp5.duration_min AS dp5_duration
, dp4.session_count AS dp4_session_count
, dp5.session_count AS dp5_session_count
, dp4.avg_session_duration AS dp4_avg_session_duration
, dp5.avg_session_duration AS dp5_avg_session_duration
, dp4_dr.detection_rate AS dp4_detection_rate
, dp5_dr.detection_rate AS dp5_detection_rate
FROM all_metrics dp4
JOIN all_metrics dp5
  ON dp4.tvid = dp5.tvid
 AND dp4.time_period = dp5.time_period
 AND dp4.session_type = dp5.session_type
JOIN detection_rate dp4_dr
  ON dp4.tvid = dp4_dr.tvid
 AND dp4.time_period = dp4_dr.time_period
JOIN detection_rate dp5_dr
  ON dp5.tvid = dp5_dr.tvid
 AND dp5.time_period = dp5_dr.time_period
JOIN both_tp_filter f
  ON f.tvid = dp4.tvid
WHERE dp4.pipeline = 'DP4'
  AND dp5.pipeline = 'DP5'
  AND dp4_dr.pipeline = 'DP4'
  AND dp5_dr.pipeline = 'DP5'
GROUP BY ALL
ORDER BY 1, 2, 3
LIMIT 1000
```
